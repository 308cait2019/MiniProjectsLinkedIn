```python
from pathlib import Path
import rasterio
import simplekml
import matplotlib.pyplot as plt
```

```python
# Extract float folder of tif imagery collection exported from GEE
img_dir = Path("C:/Users/308ca/Desktop/ProjectsGEE/Tornadoes4172026_4182026/Goes-2-TornadoProj-41926/GEE05292026CloudsIRImageCollection-1-Channel-Float")
kml = simplekml.Kml()

print("Processing original float GeoTIFFs and rendering to PNG")

# Loop through all sorted .tif files
for file_path in sorted(img_dir.glob("*.tif")):
    # Read spatial boundaries and single data channel
    with rasterio.open(file_path) as src:
        w, s, e, n = src.bounds
        data = src.read(1)  # Read band 1 for single channel float

    # Extract timestamp from filename
    time_part = file_path.stem.split("_")[-1]  
    date_part = file_path.stem.split("_")[-2]  
    timestamp = f"{date_part}T{time_part[:2]}:{time_part[2:]}:00Z"

    # Save a standard web-ready PNG version using a grayscale colormap like visualized in GEE 
    png_path = file_path.with_suffix('.png')
    plt.imsave(png_path, data, cmap='gray')

    # Create Ground Overlay linked to new PNG file
    ground = kml.newgroundoverlay(name=png_path.name)
    ground.icon.href = png_path.name  # Tells Google Earth to look right next to it
    ground.timespan.begin = timestamp  # Activates the Time Slider
    
    # Coordinates
    ground.latlonbox.north, ground.latlonbox.south = n, s
    ground.latlonbox.east, ground.latlonbox.west = e, w

# Save KML file inside same folder as images to work (same directory)
kml_output = img_dir / "GOES19_Timeline.kml"
kml.save(str(kml_output))

print(f"\nWorked: Generated PNG files and saved KML to: {kml_output}")
print("Double-click new KML file to open Google Earth Pro time slider.")
```

```python
# Extract folder of RGB tif imagery collection exported from GEE
img_dir = Path("C:/Users/308ca/Desktop/ProjectsGEE/Tornadoes4172026_4182026/Goes-2-TornadoProj-41926/GEE06122026BTDCloudFocusCollection-RGB")
kml = simplekml.Kml()

print("Processing RGB GeoTIFFs and rendering to PNG")

# Loop through all sorted .tif files
for file_path in sorted(img_dir.glob("*.tif")):
    # Read spatial boundaries and all data channels (RGB)
    with rasterio.open(file_path) as src:
        w, s, e, n = src.bounds
        data = src.read().transpose(1, 2, 0)  # Handles the 3 bands exported by .visualize()

    # Extract timestamp from filename (Kept exactly like your original logic)
    time_part = file_path.stem.split("_")[-1]  
    date_part = file_path.stem.split("_")[-2]  # This now pulls '2026-04-17' directly from GEE
    timestamp = f"{date_part}T{time_part[:2]}:{time_part[2:]}:00Z"

    # Save a standard web-ready PNG version (Uses GEE's baked-in visualization colors)
    png_path = file_path.with_suffix('.png')
    plt.imsave(png_path, data)

    # Create Ground Overlay linked to new PNG file
    ground = kml.newgroundoverlay(name=png_path.name)
    ground.icon.href = png_path.name  
    
    # Set time constraints to trigger the Google Earth Pro Time Slider
    ground.timespan.begin = timestamp  
    ground.timespan.end = timestamp    

    # Coordinates
    ground.latlonbox.north, ground.latlonbox.south = n, s
    ground.latlonbox.east, ground.latlonbox.west = e, w

# Save KML file inside same folder as images to work (same directory)
kml_output = img_dir / "GOES19_Timeline_Clouds_RGB.kml"
kml.save(str(kml_output))

print(f"\nWorked: Generated PNG files and saved KML to: {kml_output}")
print("Double-click new KML file to open Google Earth Pro time slider.")
```
